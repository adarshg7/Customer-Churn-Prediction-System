```python
def group_reasons(reason):
    if pd.isna(reason): return 'Active'
    reason = str(reason).lower()
    if 'competitor' in reason: return 'Competition'
    if 'price' in reason or 'charge' in reason: return 'Cost'
    if 'dissatisfaction' in reason or 'poor' in reason or 'reliability' in reason: return 'Service Quality'
    if 'moved' in reason or 'deceased' in reason: return 'Personal'
    return 'Other'
df['Churn Reason Grouped'] = df['Churn Reason'].apply(group_reasons)
```

```python
from sklearn.model_selection import train_test_split, RepeatedStratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.ensemble import RandomForestClassifier, StackingClassifier
from xgboost import XGBClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import classification_report, roc_auc_score
```

```python
leakage_cols = [
    'Churn Label', 
    'Churn Score', 
    'CLTV', 
    'CustomerID', 
    'Count', 
    'Zip Code'
]
reason_cols = [col for col in df.columns if 'Reason_' in col]

X = df.drop(columns=leakage_cols + reason_cols + ['Churn Value', 'Lat Long'], errors='ignore')
y = df['Churn Value']
```

```python

base_models = [
    ('xgb', XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.05, scale_pos_weight=3, random_state=42)),
    ('rf', RandomForestClassifier(n_estimators=200, max_depth=12, class_weight='balanced', random_state=42))
]

stacker = StackingClassifier(
    estimators=base_models,
    final_estimator=LogisticRegression(),
    cv=5
)

model_pipeline = ImbPipeline(steps=[
    ('scaler', StandardScaler()),
    ('smote', SMOTE(random_state=42)),
    ('stacking', stacker)
])
```

```python
final_model = CalibratedClassifierCV(model_pipeline, cv=3,method='sigmoid')

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, stratify=y, random_state=42)
final_model.fit(X_train, y_train)
```

```python
probs = final_model.predict_proba(X_test)[:,1]
preds = final_model.predict(X_test)

print(classification_report(y_test, preds))
print(f"ULTIMATE ROC-AUC: {roc_auc_score(y_test, probs):.4f}")
```
